# Plotting $\ell$ vs $C_\ell$ and $D_\ell$

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
ell, cl_obs, dl_obs = np.loadtxt("Saved Data/APS_anafast.txt", skiprows=1, unpack=True)

fig, ax = plt.subplots(1,2, figsize=(10, 6))

ax[0].plot(ell, cl_obs, label=r'$C_\ell^{\rm obs}$', color='blue')
ax[1].plot(ell, dl_obs, label=r'$D_\ell^{\rm obs}$', color='orange')

ax[0].set_xlabel(r'Multipole moment $\ell$')
ax[0].set_ylabel(r'$C_\ell$')

ax[1].set_xlabel(r'Multipole moment $\ell$')
ax[1].set_ylabel(r'$D_\ell$')

ax[0].set_title('Angular Power Spectrum - C_l')
ax[1].set_title('Angular Power Spectrum - D_l')

ax[0].grid(True, which='both', linestyle='--', linewidth=0.5)
ax[0].set_xlim(-10, 200) 
ax[1].grid(True, which='both', linestyle='--', linewidth=0.5)

plt.show()
```

### For $\ell$ vs $C_\ell$, even at range of 0 to 200, the peaks are not evident for $\ell$>150. This motivates to instead plot $\ell$ vs $D_\ell$ which is more sensitive to the peaks.

---
# Conclusion

$C_\ell$ falls steeply with $\ell$, so on a linear axis the first peak dominates and
everything past $\ell \approx 150$ is flattened into the baseline. Rescaling to

$$D_\ell = \frac{\ell(\ell+1)\,C_\ell}{2\pi}$$

cancels most of that fall-off, since $C_\ell \propto 1/\ell(\ell+1)$ for a scale
invariant spectrum. All the acoustic peaks then sit at comparable heights.

This is not only presentation. The fit is driven by peak positions and relative heights,
so working in $D_\ell$ means

$$\chi^2 = \sum_\ell \frac{\left(D_\ell^{\rm obs} - D_\ell^{\rm th}\right)^2}{\sigma_\ell^2}$$

weights the higher peaks meaningfully instead of letting the first peak dominate the sum.
Every spectrum from here on is plotted and fitted as $D_\ell$.
